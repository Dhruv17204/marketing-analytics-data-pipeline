# Silver Layer EDA Analysis
Project: Marketing Data Pipeline – Logicrest  
Objective: Perform Exploratory Data Analysis on the Silver dataset to validate data quality, identify inconsistencies, and prepare the dataset for Gold layer dimensional modeling and BI reporting.

```python
silver_df = spark.table("water_bottle_db.silver_packaged_water_marketing_clean")

print("Total rows:", silver_df.count())
print("Total columns:", len(silver_df.columns))

display(silver_df.limit(10))
```

```python
display(silver_df.describe())
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, col, desc

w = Window.partitionBy("transaction_id").orderBy(desc("silver_ingestion_ts"))

silver_dedup_df = (
    silver_df
    .withColumn("rn", row_number().over(w))
    .filter(col("rn") == 1)
    .drop("rn")
)

print("Original rows:", silver_df.count())
print("Deduplicated rows:", silver_dedup_df.count())
```

```python
silver_dedup_df.printSchema()
```

```python
import pandas as pd

schema_summary = pd.DataFrame({
    "column": silver_dedup_df.columns,
    "datatype": [f.dataType.simpleString() for f in silver_dedup_df.schema.fields]
})

display(schema_summary)
```

```python
from pyspark.sql.functions import col, count, when

null_summary = silver_dedup_df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in silver_dedup_df.columns
])

display(null_summary)
```

```python
dup_tx = (
    silver_df
    .groupBy("transaction_id")
    .count()
    .filter("count > 1")
)

print("Duplicate transactions:", dup_tx.count())
display(dup_tx.limit(20))
```

```python
sanity_checks = silver_dedup_df.selectExpr(
"count(*) as total_rows",
"sum(case when selling_price > mrp then 1 else 0 end) as selling_price_gt_mrp",
"sum(case when cost_price > selling_price then 1 else 0 end) as cost_price_gt_selling",
"sum(case when sales_units <= 0 then 1 else 0 end) as invalid_sales_units",
"sum(case when gross_revenue < 0 then 1 else 0 end) as negative_revenue",
"sum(case when gross_profit < 0 then 1 else 0 end) as negative_profit"
)

display(sanity_checks)
```

```python
display(
    silver_dedup_df
    .groupBy("brand_std")
    .count()
    .orderBy("count", ascending=False)
)
```

```python
display(
    silver_dedup_df
    .groupBy("region","state","city")
    .count()
    .orderBy("count", ascending=False)
)
```

```python
import matplotlib.pyplot as plt

pdf = silver_dedup_df.select("sales_units").toPandas()

plt.figure()
pdf["sales_units"].hist(bins=30)
plt.title("Sales Units Distribution")
plt.xlabel("Sales Units")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Select revenue and cast to float for plotting
pdf_revenue = silver_dedup_df.select(col("gross_revenue").cast("float")).toPandas()

# 2. Create the Histogram
plt.figure(figsize=(10, 6))
plt.hist(pdf_revenue["gross_revenue"], bins=50, color='#2ecc71', edgecolor='black', alpha=0.7)

# 3. Add formatting
plt.title("Distribution of Gross Revenue", fontsize=14, fontweight='bold')
plt.xlabel("Gross Revenue ($)", fontsize=12)
plt.ylabel("Number of Transactions", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
import seaborn as sns

pdf = silver_dedup_df.select(
"sales_units",
"selling_price",
"marketing_spend",
"gross_revenue",
"gross_profit"
).toPandas()

corr = pdf.corr()

plt.figure()
sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Prepare data - casting marketing_spend to float for plotting
pdf_mkt = silver_dedup_df.select(
    col("marketing_spend").cast("float"), 
    "campaign_type"
).toPandas()

# 2. Create the Box Plot
plt.figure(figsize=(12, 6))
sns.boxplot(x='campaign_type', y='marketing_spend', data=pdf_mkt, palette="Set3")

# 3. Add styling for a professional report look
plt.title("Distribution of Marketing Spend by Campaign Type", fontsize=14, fontweight='bold')
plt.xlabel("Campaign Type", fontsize=12)
plt.ylabel("Marketing Spend ($)", fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Get the null counts and convert to a format suitable for plotting
null_data = null_summary.toPandas().transpose().reset_index()
null_data.columns = ['Column', 'Null_Count']

# 2. Filter to only show columns that actually have nulls (to keep the graph clean)
null_data = null_data[null_data['Null_Count'] > 0].sort_values(by='Null_Count', ascending=False)

# 3. Create the plot
plt.figure(figsize=(10, 6))
sns.barplot(x='Null_Count', y='Column', data=null_data, palette='Reds_r')
plt.title("Missing Value Analysis by Column", fontsize=14, fontweight='bold')
plt.xlabel("Number of Missing Values")
plt.ylabel("Column Name")
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.show()
```

```python
# Average revenue by campaign type

campaign_rev = (
    silver_dedup_df
    .groupBy("campaign_type")
    .avg("gross_revenue")
    .orderBy("avg(gross_revenue)", ascending=False)
    .toPandas()
)

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(campaign_rev["campaign_type"], campaign_rev["avg(gross_revenue)"])
plt.title("Average Revenue by Campaign Type")
plt.xlabel("Campaign Type")
plt.ylabel("Average Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average sales units by product tier

tier_sales = (
    silver_dedup_df
    .groupBy("product_tier")
    .avg("sales_units")
    .orderBy("avg(sales_units)", ascending=False)
    .toPandas()
)

plt.figure(figsize=(8,5))
plt.bar(tier_sales["product_tier"], tier_sales["avg(sales_units)"])
plt.title("Average Sales Units by Product Tier")
plt.xlabel("Product Tier")
plt.ylabel("Average Sales Units")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average rating by platform

platform_rating = (
    silver_dedup_df
    .groupBy("platform_source")
    .avg("avg_rating")
    .orderBy("avg(avg_rating)", ascending=False)
    .toPandas()
)

plt.figure(figsize=(8,5))
plt.bar(platform_rating["platform_source"], platform_rating["avg(avg_rating)"])
plt.title("Average Rating by Platform")
plt.xlabel("Platform")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```
